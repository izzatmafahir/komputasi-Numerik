```python
# Fungsi persamaan
def f(x):
    return x**3 - x - 2
```

```python
# Interval awal
a = 1
b = 2

# Toleransi error
toleransi = 0.0001

# Maksimum iterasi
maks_iterasi = 100
```

```python
# Interval awal
a = 1
b = 2

# Toleransi error
toleransi = 0.0001

# Maksimum iterasi
maks_iterasi = 100
```

```python
if f(a) * f(b) >= 0:
    print("Interval tidak memenuhi syarat Bisection")
else:
    print("Interval memenuhi syarat Bisection")
```

```python
if f(a) * f(b) < 0:

    for i in range(1, maks_iterasi + 1):

        c = (a + b) / 2

        print(f"Iterasi {i}: a={a:.6f}, b={b:.6f}, c={c:.6f}")

        if abs(f(c)) < toleransi or abs(b-a)/2 < toleransi:
            break

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
```
