```python
import numpy as np
```

```python
# Koefisien variabel x, y, dan z
A = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
])

# Nilai hasil persamaan
B = np.array([10, 12, 13])
```

```python
print("SISTEM PERSAMAAN LINEAR TIGA VARIABEL")
print("-------------------------------------")

print("2x + y + z = 10")
print("x + 2y + z = 12")
print("x + y + 2z = 13")
```

```python
# Menyelesaikan sistem persamaan
hasil = np.linalg.solve(A, B)

# Mengambil nilai setiap variabel
x = hasil[0]
y = hasil[1]
z = hasil[2]
```

```python
print("HASIL PENYELESAIAN SPLTV")
print("------------------------")

print(f"Nilai x = {x:.2f}")
print(f"Nilai y = {y:.2f}")
print(f"Nilai z = {z:.2f}")
```

```python
# Menghitung kembali hasil persamaan
cek = np.dot(A, hasil)

print("PEMERIKSAAN HASIL")
print("-----------------")

print("Hasil perhitungan:", cek)
print("Hasil seharusnya :", B)

if np.allclose(cek, B):
    print("Kesimpulan: Hasil perhitungan BENAR")
else:
    print("Kesimpulan: Hasil perhitungan SALAH")
```
